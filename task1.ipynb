```python
from typing import List, TypedDict
from langgraph.graph import StateGraph, END, START
```

```python
class AgentState(TypedDict):
    messages: str

def greeting_agent(state: AgentState) -> AgentState:
    """Greet the user by name."""
    name = state["messages"].strip()
    state["messages"] = f"Hey {name}, how are you doing today?"
    return state

def farewell_agent(state: AgentState) -> AgentState:
    """Add a short closing after the greeting."""
    state["messages"] = state["messages"] + " Goodbye, you did an amazing job."
    return state
```

```python
graph = StateGraph(AgentState)

graph.add_node("greeting", greeting_agent)
graph.add_node("farewell", farewell_agent)

graph.set_entry_point("greeting")
graph.set_finish_point("farewell")

graph.add_edge("greeting", "farewell")

app= graph.compile()
```

```python
result= app.invoke({"messages": "Nishank"})
```
